```python
%timeit
# Посмотрим процентное соотношение жанров в нашей выборке
total_genres=sum(dM.dist_by_genres().values())
genres_share={}    

for k,v in dM.dist_by_genres().items():
    genres_share[k]=round(float((v/total_genres)*100), 2)

genres_share
```
